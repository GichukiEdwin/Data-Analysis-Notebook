```python
cars = sns.load_dataset('mpg')
cars.head()
```

```python
cars.shape
```

```python
cars = cars.dropna()
cars.shape
```

### Intro

#### Basics

```python
sns.set_style('whitegrid')
```

```python
cars.cylinders.value_counts()
```

#### Filter down to only cars that have an even number of cylinders

```python
cars = cars[cars.cylinders.isin([4,6,8])]
cars.cylinders.value_counts()
```

```python
sns.boxplot(cars.mpg)
```

```python
cars.mpg.describe()
```

```python
sns.boxplot(x=cars.origin, y=cars.mpg)
```

```python
sns.boxplot(x='origin', y='mpg', data=cars)
```

```python
sns.boxplot(x='origin', y='mpg', hue='cylinders', data=cars)
```

```python
sns.boxplot(x=cars.origin, y=cars.mpg, hue=cars.cylinders)
```

```python
cars.model_year.describe()
```

```python
cars['newer_model'] = cars.model_year > 76
cars.head()
```

```python
sns.boxplot(x='origin', y='mpg', hue='newer_model', data=cars)
```

# Styling

```python
sns.boxplot(x='mpg', y='origin', hue='cylinders', data=cars)
```

```python
sns.boxplot(x='mpg', y='origin', hue='cylinders', data=cars, order=['japan','europe','usa'])
```

```python
sns.boxplot(x='mpg', y='origin', hue='newer_model', hue_order=[True,False],order=['japan','europe','usa'], data=cars)
```

```python
sns.boxplot(x='mpg', y='origin', hue='newer_model', hue_order=[True,False],order=['japan','europe','usa'], palette='dark:green', data=cars)
```

```python
sns.boxplot(x='mpg', y='origin', hue='cylinders', hue_order=[4,6,8],order=['japan','europe','usa'], linewidth=3, width=0.5, data=cars)
```

```python
sns.boxplot(x='mpg', y='origin', order=['japan','europe','usa'], whis=0.5, data=cars)
```

```python
sns.boxplot(x='mpg', y='origin', order=['japan','europe','usa'], whis=0.5, fliersize=1, data=cars)
```

```python
sns.boxplot(x='mpg', y='origin', order=['japan','europe','usa'], whis=0.5, fliersize=1, showcaps=False, data=cars)
```
