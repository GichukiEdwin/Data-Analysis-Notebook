#### counting time zones in pure python
- suppose we are interested in the most often occurring time zones in the data set(the tz field).
- there are many ways we could do that.

##### extract a list of time zones using a smart list

````python
timezones = [rec['tz'] for rec in records if 'tz' in rec]
````

````python
timezones[:10]
````

##### producing counts by time zones using a dict to store counts while we iterate through the time zones

````python
def get_counts(sequence):
    counts = {}
    for x in sequence:
        if x in counts:
            counts[x] += 1
        else:
            counts[x] = 1
    return counts
````

##### using python library to write the same fn more briefly

````python
from collections import defaultdict

def get_counts2(sequence):
    counts = defaultdict(int) # values will be initialized to zero
    for x in sequence:
        counts[x] += 1
    return counts
````

````python
counts = get_counts(timezones)
counts['America/New_York']
````

````python
len(timezones)
````

##### get the top 10 time zones and their counts

````python
def top_counts(count_dict, n=10):
    value_key_pairs = [(count, tz) for tz, count in count_dict.items()]
    value_key_pairs.sort()
    return value_key_pairs[-n:]
````

````python
top_counts(counts)
````

##### a bit of munging to fill in a substitute value for unknown and missing time zone data in the records
- the *fillna* function can replace missing (NA) values and unknown (empty strings) values can be replaced by boolean array indexing

##### the digital equivalent of wiping fingerprints off your data before analysis 👀✨

````python
clean_tz = df['tz'].fillna('Missing')
````

### 🧠
##### because missing values are like potholes on a data highway, they can mess things up when you’re counting, plotting, or analyzing.
##### replacing them with 'Missing' gives you a clean label to group or count later.



##### some time zones might not be missing (null) but just empty
##### like this: ```{'tz': ''}  ← not NaN, just blank```

````python
clean_tz[clean_tz == ''] = 'Unknown'
````

##### finds entries that are empty strings ('' — not null, just blank).
##### replaces those with 'Unknown'.

#### No NaNs. No empty strings. Just facts 🔥

````python
tz_counts = clean_tz.value_counts()
````

````python
tz_counts[:10]
````

##### making a plot of this data using a plotting library

````python
tz_counts[:10].plot(kind='barh', rot=0, title='Top time zones in the 1.usa.gov sample data')
````

this a smart list or a list comprehension - a one-liner loop
- for every value `x` in that column, ```.split()``` splits the string into words based on spaces
- ```.split()[0]``` grabs the first word from that list or the thing before the first space
- so it is basically saying: 'For every non-null value in column a, split the string, and keep just the first word.'

##### ```df.a.notnull()``` - returns a boolean series with a `True` value for rows where the user agent *a* is not __NaN__ or **Not missing** and a `False` value where it is missing
##### ```df[df.a.notnull()]``` - filters the rows to keep only the rows with non-missing user agents
##### 🧠 `c_df` is a clean version DataFrame of the original DataFrame that contains only rows where user agents are present, so no mystery data 😎

````python
c_df = df[df.a.notnull()]
````

##### 💡 computing a value whether each row is Windows or not
+ ```c_df['a'].str.contains('Windows')``` - creates a boolean Series that is `True` if the string **'Windows'** apppears  in the user agent and **'False'** if it doesn't appear
+ ```np.where(condition, x, y)``` - if the above condition returns true, then assign **'Windows'** else assign **'Not Windows'**
+ we get a numpy array called **os** with values like `['Windows', 'Not Windows', 'Windows', 'Not Windows', 'Windows']`
+ ```array(['Windows', 'Not Windows', 'Windows', 'Not Windows', 'Windows'],dtype='<U11')```

````python
import numpy as np
````

````python
os = np.where(c_df['a'].str.contains('Windows'), 'Windows', 'Not Windows')
````

````python
os[:5]
````

##### grouping the cleaned DataFrame `c_df` by it time zone column and this new list of operating systems
+ `by_tz_os = c_df.groupby(['tz', os])`
+ **'tz'** - the time zone column
+ **os** - the **os** label we created above(**'Windows'** or **'Not Windows'**)
+ even though *os* is not a column in `c_df`, ***Pandas*** is smart, as long as it’s the same length as c_df, it treats it like a virtual column when grouping. 🚀

````python
by_tz_os = c_df.groupby(['tz', os])
````

`agg_counts = by_tz_os.size().unstack().fillna(0)`
+ `.size()` - counts how many rows are in each group (ie. for each `'tz'` + **os** pair)
+ `.unstack()` - pivots the **os** categories out of the index and into columns, making a 2D table (DataFrame)
+ `.fillna()` - replaces any missing values, (like if a time zone has only Windows users and no Non Windows users) with 0
+ `agg_counts = by_tz_os.size().unstack().fillna(0)` - creates a clean table showing user count per OS per time zone

````python
agg_counts = by_tz_os.size().unstack().fillna(0)
````

````python
agg_counts[:10]
````

````python
agg_counts.shape
````

````python
indexer = agg_counts.sum(1).argsort()
````

````python
indexer[:10]
````

````python
count_subset = agg_counts.take(indexer)[-10:]
count_subset
````

````python
count_subset.plot(kind='barh', stacked=True, title='Top time zones by Windows and non-Windows users')
````

````python
normed_subset =count_subset.div(count_subset.sum(1), axis=0)
normed_subset.plot(kind='barh', stacked=True, title='Percentage Windows and non-Windows users in top-occurring timezones')
````
